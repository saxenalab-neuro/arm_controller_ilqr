# Alignment Index Across Feedback Conditions

Loads the saved alignment index for each feedback condition and displays the cross-condition comparison.

## 1. Imports

Load the numerical and plotting dependencies used by the analysis.

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
```

## 2. Environment Sanity Check

Check tensor slicing before loading analysis results.

```python
a= torch.FloatTensor([1, 2, 3])
print(a[:2].shape)
```

## 3. Load Saved Alignment Indices

Read the scalar outputs produced by the four orthogonality notebooks.

```python
align_index= np.load('./align_index.npy')[0]
align_index_p= np.load('./align_index_p.npy')[0]
align_index_v= np.load('./align_index_v.npy')[0]
align_index_h= np.load('./align_index_h.npy')[0]
```

## 4. Plot the Cross-Condition Comparison

Display complete-state, proprioceptive, visual, and recurrent feedback conditions.

```python
plt.figure(figsize= (10, 5))

sf= ['Complete State \n Feedback', 'Proprioception Only', 'Vision Only', 'Recurrence Only']
align_idxs= [align_index, align_index_p, align_index_v, align_index_h]

plt.xlabel('Feedback')
plt.ylabel('Alignment Index')
plt.title('Alignment Index (OFC)')


plt.bar(sf, align_idxs, alpha= 0.8)


plt.show()
```
